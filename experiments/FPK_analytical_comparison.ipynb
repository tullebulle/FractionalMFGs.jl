# Verification of FPK-Solve

Consider the fractional FPK equation
\begin{align}
\begin{cases}
    \partial_t m(x,t) -\nu (-\Delta)^{\frac{\alpha}{2}} m - \partial_x (mv) = 0 \quad  & \text{in} \quad T \times (0, T) \\ 
    \int_{\mathbb{T}^2} m(x,t) \, dx = 1, \quad m \geq 0, \quad & \text{in} \quad (0, T)\\
    m(x, T) = m_T(x) \in P(T^d) , \qquad u(\cdot, 0) = G[m(\cdot, 0)] \qquad &\text{in} \quad T^d. 
\end{cases}
\end{align}

\begin{align}
    \partial_t m(x,t) -\nu (-\Delta)^{\frac{\alpha}{2}} m  &=  \partial_x (mv)  \\ 
        \frac{1}{m(x, t)} \int_0^x \bigg( \partial_t m(r,t) -\nu (-\Delta)^{\frac{\alpha}{2}} m(r, t) \bigg) dr  &= v(x, t),   \\ 
\end{align}
and furthermore, $v = - \nabla_p H(x, u_x) = - \nabla_p |u_x|^2 = - 2 u_x \implies u = -\frac12 \int v dx.$ So, after we have found the velocity $v$, we would need to find the $u$ satifying this, so that we don't need to change the code that much.

We found that for 
$$
m(x,t) = Z \cos^2(2\pi x - c_1t^2) + \epsilon
$$
where $Z$ is a normalization constant. We found that for $\nu = 0.1, \alpha = 1.0$,
$$
\nu(-\Delta)^{\frac{\alpha}{2}} m(x,t) = c_0 \cos(4\pi x - 2c_1t^2)
$$
$$
\partial_t m(x,t) = \partial_t Z \cos^2(2\pi x - c_1t^2) = Z 2 \cos(2\pi x - c_1t^2) (-\sin(2\pi x - c_1t^2)) (-2c_1t) = Z c_1 4t \cos(2\pi x - c_1t^2) \sin(2\pi x - c_1t^2).
$$

Notice further that 
$$
\partial_x m(x,t) = -Z 4\pi \cos(2\pi x - c_1t^2) \sin(2\pi x - c_1t^2) = \frac{-\pi}{c_1 t} \partial_t m(x, t).
$$

Hence, 
$$
\int_0^x  \partial_t m(r,t) dr = \int_0^x  \frac{-c_1 t}{\pi} \partial_x m(r,t) dr = \frac{-c_1 t}{\pi}[ m(x,t) - m(0, t)],
$$
and
$$
\int_0^x \nu (-\Delta)^{\frac{\alpha}{2}} m(r, t) dr = \int_0^x c_0 \cos(4\pi r - 2c_1t^2) dr = \frac{c_0}{4 \pi }[\sin(4\pi x + 2c_1t^2) - \sin(2c_1t^2)]
$$

$$
\int_0^x \big( \partial_t m(r,t) +  \nu (-\Delta)^{\frac{\alpha}{2}} m(r, t) \big)  dr = \frac{Zt}{\pi}[ \cos^2(c_1t^2) -  \cos^2(2\pi x - c_1t^2) ] - \frac{c_0}{4 \pi }[\sin(4\pi x - 2c_1t^2) + \sin(2c_1t^2)] 
$$

Hence, we get

\begin{align*}
 v(x, t)  &= \frac{1}{m(x, t)} \int_0^x \bigg( \partial_t m(r,t) -\nu (-\Delta)^{\frac{\alpha}{2}} m(r, t) \bigg) dr   \\ 
&= \frac{1}{Z \cos^2(2\pi x - c_1t^2) + \epsilon} \bigg(\frac{-c_1t}{\pi}[ m(x,t) - m(0, t)] - \frac{c_0}{4 \pi }[\sin(4\pi x - 2c_1t^2) - \sin(- 2c_1t^2)] \bigg)   \\
&= \frac{1}{Z \cos^2(2\pi x - c_1t^2) + \epsilon} \bigg(\frac{-c_1t}{\pi}[ Z \cos^2(2\pi x - c_1t^2) + \epsilon - Z \cos^2(-c_1t^2) - \epsilon] - \frac{c_0}{4 \pi }[\sin(4\pi x - 2c_1t^2) - \sin(- 2c_1t^2)] \bigg)  \\
&= \frac{\frac{c_1tZ}{\pi} \bigg( \cos^2(c_1t^2) -  \cos^2(2\pi x - c_1t^2) \bigg) - \frac{c_0}{4 \pi }[\sin(4\pi x - 2c_1t^2) + \sin(2c_1t^2)] }{Z \cos^2(2\pi x - c_1t^2) + \epsilon}  
\end{align*}
Since
$$
u(x,t) = u(0,t)  -\frac12 \int_0^x v(r, t) dr
$$

```julia
using QuadGK, Plots, LaTeXStrings
using LinearAlgebra;
using SpecialFunctions: gamma, loggamma, zeta
```

```julia
function velocity(x, t) #this is now correct
    int_m_t = -c_1*t/pi * (m_func(x, t) - m_func(0, t)) #minus comes from the factor -c_1 t / pi
    int_mLP = c_0 / (4*pi) * (sin(4*pi*x - c_1*2*t^2) + sin(c_1*2*t^2)) #+ c_2*x #
    m = m_func(x, t)
    return -(int_m_t + ν*int_mLP) / m
end

function partial_x_matrix(mat, h) # take in N_h x N_T matrices
    ddx_mat = Array{Float64}(undef, N_h, N_T);
    for i in 1:N_T
        ddx_mat[:, i] = ddx_torus(mat[:, i], h)
    end
    return ddx_mat
end

function ddx_torus(vec, h)
    return [diff(vec) / h ; (vec[1] - vec[end]) / h]
end

function ddx_torus_central(vec, h)
    v1 = [diff(vec) / h ; (vec[1] - vec[end]) / h]
    v2 = [(vec[1] - vec[end]) / h; diff(vec) / h ]
    return (v1 + v2) ./ 2
end

function partial_x_matrix_central(mat, h) # take in N_h x N_T matrices
    ddx_mat = Array{Float64}(undef, N_h, N_T);
    for i in 1:N_T
        ddx_mat[:, i] = ddx_torus_central(mat[:, i], h)
    end
    return ddx_mat
end

function plus(x)
    return max(x, 0)
end

function minus(x)
    return max(-x, 0)
end


function create_U(v_arr)
    U = Array{Float64}(undef, N_h, N_T)
    for n in 1:N_T
        cumsumm = 0
        for i in 1:N_h
            cumsumm += h*v_arr[i, n]
            U[i, n] = cumsumm
        end
    end
    return U
end


function create_D_vecs(u_vec, h) # returns Dplusminus and Dminusplus vectors
    Dplus = [diff(u_vec); 0] ./ h
    Dminus = [0; diff(u_vec)] ./ h
    Dplus[end] = (Dplus[end-1] + Dplus[1]) / 2
    Dminus[1]= (Dplus[end-1] + Dplus[1]) / 2
    ###
    Dplus = Dminus # this is since we create the derivative from u and work backwards
    ###
    Dplusminus = minus.(Dplus)
    Dminusplus = plus.(Dminus)
    Dplusplus = plus.(Dplus)
    Dminusminus = minus.(Dminus)
    return (Dplusminus, Dminusplus, Dplusplus, Dminusminus)
end

function TS(m, α, N_h, R)
    if m==0
        return 0
    end
    K_α = 0.0
    for ν in -R:1:R
        K_α += exp( loggamma( abs(m - N_h*ν) - α/2 )  - loggamma( ( abs(m - N_h*ν) + 1 + α/2) )  )
    end
    return K_α
end

function create_DPL_matrix(N_h, α, h, R)
    DPL_mat = Matrix{Float64}(undef, N_h, N_h)
    zeta_term= zeta(1+α)
    for i in 1:N_h
        for j in 1:N_h
            DPL_mat[i,j] = TS(abs(i-j), α, N_h, R)
        end
    end
    for γ in 1:N_h
        DPL_mat[γ, γ] = -(sum([TS(β-γ, α, N_h, R) for β in 1:N_h]) + 2*zeta_term - 
        sum([1/k^(1+α) for k in 1:((R+1)*N_h - γ) ]) - sum([1/k^(1+α) for k in 1:(R*N_h + γ - 1) ]))
    end
    c_α = 2^α * gamma((1+α)/2) / (√π * abs(gamma(-α/2)) )
    return -c_α / (h^α) * DPL_mat
end

function tranport_matrix(u_vec, N_h, h) # modify to avoid integrating the goshdarn v.
    (DPM, DMP, DPP, DMM) = create_D_vecs(u_vec, h);
    TU_mat = zeros(N_h, N_h)
    tridiag = Tridiagonal(DMM[1:end-1], -(DMP + DPM), DPP[2:end])
    TU_mat[1, N_h] = DMM[end]
    TU_mat[N_h, 1] = DPP[1]
    TU_mat += tridiag
    TU_mat /= h # not dividing by h
    return TU_mat
end

function FPK_step_forward(U_np1, M_n, DPL_mat, N_h, Δt, h, ν, α, R) # not exactly verified, but DPL and TU is verified.
    TU = tranport_matrix(U_np1, N_h, h)
    total_mat = 1.0I(N_h) + Δt *(ν*DPL_mat + TU)
    M_np1 = total_mat \ M_n
    return M_np1
end

function FPK_solve_forward(u_arr, M_0, N_h, Δt, N_T, h, ν, α, R)
    M_mat = Array{Float64}(undef, N_h, N_T)
    M_mat[:, 1] = M_0;
    DPL_mat = create_DPL_matrix(N_h, α, h, R)
    for n in 1:(N_T-1)
        M_mat[:, n+1] = FPK_step_forward(u_arr[:, n+1], M_mat[:, n], DPL_mat, N_h, Δt, h, ν, α, R)
    end
    return M_mat
end
```

```julia
using QuadGK
# creating m(x,t)
h=0.001
x_vec = 0:h:(1-h)
N_h = length(x_vec)
Δt = 0.005
α = 1.5
t_vec = 0:Δt:(2-Δt)
N_T = length(t_vec)
ν = 0.1
ϵ = 0.2 #to avoid singular velocities
c_1 = 0.4


m_func_unnorm = (x) -> cos(2*pi*x)^2
C_int, error = quadgk(m_func_unnorm, 0, 1);
Z = (1-ϵ)/C_int
m_func = (x, t) -> Z*cos(2*pi*x - c_1*t^2)^2 + ϵ
m_arr = [m_func(x, t) for x in x_vec, t in t_vec]; # exact solution


#DPL
DPL_mat = create_DPL_matrix(N_h, α, h, 20)
m_initial = m_arr[:, 1]
FL_initial = DPL_mat * m_initial; 
k_1 = maximum(FL_initial);
k_2 = minimum(FL_initial); c_2 = (k_1 + k_2)/2; c_0 = (k_1 - k_2)/2;
FL_mat = DPL_mat * m_arr
surrogate_func = (x, t) -> c_0*cos(4*pi*x - 2*c_1*t^2) + c_2
surrogate_FL = [surrogate_func(x, t) for x in x_vec, t in t_vec] # creating a surrogate

#sample for comparison plot
t_n_index = 1 #Int(length(t_vec)/2)
m_sample = m_arr[:, t_n_index]
surro_sample = surrogate_FL[:, t_n_index]
FL_sample = DPL_mat * m_sample;
```
